2. Run the cell below. The widget runs **in the browser** (Jupyter Lab) and opens `ws://127.0.0.1:8000/ws/chat`.

Optional query params: `?customer_id=CUS123`.

````python
from IPython.display import HTML, display

WS_URL = "ws://127.0.0.1:8000/ws/chat"
CUSTOMER_ID = "CUS123"

display(HTML(f"""
<style>
  #chat-box {{
    font-family: ui-sans-serif, system-ui, sans-serif;
    max-width: 720px;
    border: 1px solid #ccc;
    border-radius: 8px;
    padding: 12px;
  }}
  #chat-log {{
    height: 320px;
    overflow-y: auto;
    white-space: pre-wrap;
    background: #f7f7f7;
    padding: 8px;
    margin-bottom: 8px;
  }}
  #chat-row {{ display: flex; gap: 8px; }}
  #chat-input {{ flex: 1; padding: 8px; }}
</style>
<div id="chat-box">
  <div id="chat-status">Disconnected</div>
  <div id="chat-log"></div>
  <div id="chat-row">
    <input id="chat-input" placeholder="Describe a disputed transaction..." />
    <button id="chat-send">Send</button>
  </div>
</div>
<script>
(() => {{
  const url = new URL({WS_URL!r}, window.location.href);
  url.searchParams.set("customer_id", {CUSTOMER_ID!r});
  const wsUrl = url.toString();
  const log = document.getElementById("chat-log");
  const status = document.getElementById("chat-status");
  const input = document.getElementById("chat-input");
  const sendBtn = document.getElementById("chat-send");
  const write = (who, text) => {{
    log.textContent += who + ": " + text + "\\n\\n";
    log.scrollTop = log.scrollHeight;
  }};
  const socket = new WebSocket(wsUrl);
  socket.onopen = () => {{ status.textContent = "Connected " + wsUrl; }};
  socket.onclose = () => {{ status.textContent = "Disconnected"; }};
  socket.onerror = () => {{ status.textContent = "Error — is factored-api running on :8000?"; }};
  socket.onmessage = (event) => {{
    let payload = event.data;
    try {{ payload = JSON.parse(event.data); }} catch (_err) {{}}
    if (payload && payload.reply) {{
      write("agent", payload.reply + (payload.sql ? "\\nSQL: " + payload.sql : ""));
    }} else {{
      write("agent", typeof payload === "string" ? payload : JSON.stringify(payload, null, 2));
    }}
  }};
  const send = () => {{
    const message = input.value.trim();
    if (!message || socket.readyState !== WebSocket.OPEN) return;
    socket.send(JSON.stringify({{ message }}));
    write("you", message);
    input.value = "";
  }};
  sendBtn.onclick = send;
  input.addEventListener("keydown", (event) => {{ if (event.key === "Enter") send(); }});
}})();
</script>
"""))
````

Try this after the session starts:

1. `Hi, I noticed a fraudulent transaction on September 30th. The amount was around 35 dollars.`
2. Then describe one of your last 3 transactions (date / amount / merchant) when the agent asks.
3. Try a blocked prompt such as `delete the transactions table`.
